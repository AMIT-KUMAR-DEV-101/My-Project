```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("E-commerce.csv")
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.shape
df.size
df.duplicated().sum()
df.describe(include='all')
```

```python
df["Delivery_charge"]=df["Delivery_charge"].fillna(df["Delivery_charge"].mean())
df
df["Sales"]=df["Sales"].fillna(df["Sales"].mean())
df
```

```python
df.columns=df.columns.str.strip()
df
```

```python
df.columns=df.columns.str.replace(' ','',regex=False)
df
```

```python
df["Age_category"]=df["Age_category"].str.strip().str.title()
df
```

```python
text_columns = df.select_dtypes(
    include='object'
).columns

for col in text_columns:
    df[col] = df[col].str.strip()
df
```

```python
date_columns=[
    'Order_date',
    'Shipping_date',
    'Delivered_date']
for col in date_columns:
    df[col]=pd.to_datetime(
        df[col],
        errors='coerce')
df
```

```python
Category_Sales=df.groupby('Category')['Sales'].sum()
print(Category_Sales)
```

```python
Region_Profit=df.groupby('Region')["Profit"].mean()
print(Region_Profit)
```

```python
payment_analysis = df.groupby('Payment_method').agg({'Sales':'sum','Profit':'sum','Customer_id':'count'})
payment_analysis
```

```python
gender_analysis = df.groupby('Gender').agg({
    'Sales': 'mean',
    'Profit': 'mean',
    'Customer_rating': 'mean'
})

gender_analysis
```

```python
Total_Profit=df["Profit"].sum()
print(Total_Profit)
```

```python
total_sales = df['Sales'].sum()

total_profit = df['Profit'].sum()

average_sales = df['Sales'].mean()

average_profit = df['Profit'].mean()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Average Sales:", average_sales)
print("Average Profit:", average_profit)
```

```python
profit_margin = (total_profit / total_sales) * 100

print("Profit Margin:", profit_margin, "%")
```

```python
df.describe()
```

```python
Category_Sales.plot(
    kind='bar',
    figsize=(10, 5),
    title='Sales by Product Category'
)

plt.ylabel('Total Sales')
plt.xlabel('Category')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
Monthly_Sales = (
    df.groupby(
        df['Order_date'].dt.to_period('M')
    )['Sales'].sum()
)

Monthly_Sales.index = Monthly_Sales.index.astype(str)

Monthly_Sales.plot(
    figsize=(12, 5),
    title='Monthly Sales Trend',marker='o'
)

plt.xlabel('Month')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
Region_Profit.plot(
    kind='bar',
    figsize=(8, 5),
    title='Profit by Region'
)

plt.ylabel('Profit')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='Sales',
    bins=30,
    kde=True
)

plt.title('Sales Distribution')
plt.show()
```

```python
Category_Sales = df.groupby('Category')['Sales'].sum()

Category_Sales.plot(kind='bar', figsize=(9,5), title='Category-wise Sales')
plt.ylabel('Total Sales')
plt.show()
```

```python
sns.countplot(data=df, x='Customer_rating')

plt.title('Customer Rating Distribution')
plt.xlabel('Customer Rating')
plt.ylabel('Number of Customers')
plt.show()
```

```python
print("========== E-COMMERCE DATA CLEANING REPORT ==========")

print("\n1. ORIGINAL DATASET SHAPE:")
print(df.shape)

print("\n2. FINAL DATASET SHAPE:")
print(df.shape)

print("\n3. TOTAL MISSING VALUES:")
print(df.isnull().sum().sum())

print("\n4. TOTAL DUPLICATE ROWS:")
print(df.duplicated().sum())

print("\n5. DATA TYPES:")
print(df.dtypes)

print("\n6. TOTAL SALES:")
print(df['Sales'].sum())

print("\n7. TOTAL PROFIT:")
print(df['Profit'].sum())

print("\n8. AVERAGE SALES:")
print(df['Sales'].mean())

print("\n9. AVERAGE PROFIT:")
print(df['Profit'].mean())

print("\n10. PROFIT MARGIN:")
print((df['Profit'].sum() / df['Sales'].sum()) * 100)

print("\n11. TOP 5 PRODUCT CATEGORIES BY SALES:")
print(df.groupby('Category')['Sales'].sum().sort_values(ascending=False).head())

print("\n12. REGION-WISE PROFIT:")
print(df.groupby('Region')['Profit'].sum())

print("\n13. FINAL CLEANED DATASET:")
print(df.head())

print("\n========== REPORT COMPLETED ==========")
```
